# Test steps by case interpretation

**What it does**
Lists every case interpretation (one interpretation run over a case) and prints the test steps of one of them: first a header with the case, provider, model and date, then each step with its description, acceptance criteria and refusal criteria. It also shows them as a pandas table and as the same JSON the API returns.

**Why**
To read and review what the interpretation produced without the API or a browser. It reads through the app's own database layer, so it shows exactly what the API would return.

**How to use**
1. It reads the database chosen by `KCT_DB_DRIVER`.
   - `memory` (default): nothing is stored between processes, so this notebook only sees runs made in this same kernel, for example with `from services.app import boot; boot().interpretations.create_and_generate(["CD.pdf"], ["TP.pdf"])`. Runs made through the API are not visible here.
   - `sqlite` or `bigquery`: it sees every run, including those made through the API.
2. Open this notebook from the project's `notebooks/` folder and run the first two code cells to see the list.
3. Copy a `case_interpretation_id` from the list into the next cell and run the rest.

**Needs**: pandas for the table cell only. Terminal equivalent: `./kct.sh steps` and `./kct.sh steps <id>`.
